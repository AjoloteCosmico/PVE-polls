```python
#historial de commits
#!/usr/bin/env python3
from git import Repo
from datetime import datetime, timedelta
from collections import defaultdict

def get_commits_by_author(repo_path, days=180):
    repo = Repo(repo_path)
    since_date = datetime.now() - timedelta(days=days)
    
    author_stats = defaultdict(int)
    for commit in repo.iter_commits(since=since_date):
        author_stats[commit.author.name] += 1
    
    return author_stats

if __name__ == "__main__":
    stats = get_commits_by_author(".", days=180)
    for author, count in sorted(stats.items(), key=lambda x: x[1], reverse=True):
        print(f"{author}: {count} commits")
```

```python
import subprocess
from datetime import datetime, timedelta
from collections import defaultdict

def report_commits_by_author(days=90, repo_path='.'):
    since_date = (datetime.now() - timedelta(days=days)).strftime("%Y-%m-%d")
    cmd = [
        "git", "-C", repo_path, "log",
        f"--since={since_date}",
        "--format=%an"      # Solo el nombre del autor
    ]
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode != 0:
        print("Error ejecutando git log:", result.stderr)
        return {}
    

    
    authors = result.stdout.strip().split("\n")
    stats = defaultdict(int)
    for author in authors:
        if author:
            stats[author] += 1
    
    return stats

# Ejecutar
stats = report_commits_by_author(days=90)
for author, count in sorted(stats.items(), key=lambda x: x[1], reverse=True):
    print(f"{author}: {count} commits")
```

```python
import re
from collections import defaultdict

# Diccionario para almacenar los datos
stats = defaultdict(lambda: {'lineas': 0})

def procesar_log(archivo):
    autor_actual = ""
    with open(archivo, 'r', encoding='utf-8') as f:
        for linea in f:
            # Detectar cabecera de commit (Autor|Fecha|Hash)
            if '|' in linea and len(linea.split('|')) == 3:
                autor_actual = linea.split('|')[0]
            
            # Detectar líneas de inserciones en el formato de git --stat
            # Ejemplo: " 5 files changed, 10 insertions(+), 2 deletions(-)"
            match = re.search(r'(\d+) insertions\(\+\)', linea)
            if match:
                num_lineas = int(match.group(1))
                stats[autor_actual]['lineas'] += num_lineas

    # Mostrar resultados
    print(f"{'Colaborador':<20} | {'Total Líneas Agregadas':<20}")
    print("-" * 45)
    for autor, data in stats.items():
        print(f"{autor:<20} | {data['lineas']:<20}")

procesar_log('historial_PVE-polls (software local).txt')
```
